import pandas as pd

df = pd.read_csv("seasonal_agriculture_performance_dataset.csv")

print(df.head())
print(df.shape)

df["Rainfall_mm"] = df["Rainfall_mm"].fillna(df["Rainfall_mm"].median())
df["Soil_Moisture_pct"] = df["Soil_Moisture_pct"].fillna(df["Soil_Moisture_pct"].median())
df["Yield_Tonnes_Ha"] = df["Yield_Tonnes_Ha"].fillna(df["Yield_Tonnes_Ha"].median())

print("Missing Values After Cleaning:")
print(df.isnull().sum())

```python
print("Statistical Summary:")
print(df.describe())
```

```python
print("Unique States:")
print(df["State"].unique())

print("\nUnique Crops:")
print(df["Crop"].unique())

print("\nUnique Seasons:")
print(df["Season"].unique())

print("\nIrrigation Methods:")
print(df["Irrigation_Method"].unique())
```

```python
crop_analysis = df.groupby("Crop").agg({
    "Yield_Tonnes_Ha": "mean",
    "Production_Tonnes": "sum",
    "Revenue_INR": "mean",
    "Profit_INR": "mean"
}).round(2)

print("Crop-wise Performance:")
print(crop_analysis)
```

```python
# State-wise Performance Analysis

state_analysis = df.groupby("State").agg({
    "Yield_Tonnes_Ha": "mean",
    "Production_Tonnes": "sum",
    "Revenue_INR": "mean",
    "Profit_INR": "mean"
}).round(2)

print("State-wise Performance:")
print(state_analysis)
```

```python
# Season-wise Performance Analysis

season_analysis = df.groupby("Season").agg({
    "Yield_Tonnes_Ha": "mean",
    "Production_Tonnes": "sum",
    "Revenue_INR": "mean",
    "Profit_INR": "mean"
}).round(2)

print("Season-wise Performance:")
print(season_analysis)
```

```python
# Irrigation Method Performance Analysis

irrigation_analysis = df.groupby("Irrigation_Method").agg({
    "Yield_Tonnes_Ha": "mean",
    "Production_Tonnes": "sum",
    "Profit_INR": "mean",
    "Water_Efficiency_t_per_1000m3": "mean"
}).round(2)

print("Irrigation Method Performance:")
print(irrigation_analysis)
```

```python
# Fertilizer Usage Analysis

print("Average Fertilizer Used (kg/ha):",
      round(df["Fertilizer_kg_ha"].mean(), 2))

print("Average Yield (tonnes/ha):",
      round(df["Yield_Tonnes_Ha"].mean(), 2))

print("Average Profit (INR):",
      round(df["Profit_INR"].mean(), 2))

fertilizer_bins = pd.cut(df["Fertilizer_kg_ha"], bins=4)

fertilizer_analysis = df.groupby(
    fertilizer_bins, observed=True
).agg({
    "Yield_Tonnes_Ha": "mean",
    "Profit_INR": "mean"
}).round(2)

print("\nFertilizer Usage Group Performance:")
print(fertilizer_analysis)
```

```python
# Pesticide & Disease/Pest Risk Analysis

print("Average Pesticide Used (L/ha):",
      round(df["Pesticide_Litre_ha"].mean(), 2))

print("Average Disease/Pest Risk (%):",
      round(df["Disease_Pest_Risk_pct"].mean(), 2))

risk_analysis = df.groupby(
    pd.cut(df["Pesticide_Litre_ha"], bins=4)
).agg({
    "Disease_Pest_Risk_pct": "mean",
    "Yield_Tonnes_Ha": "mean",
    "Profit_INR": "mean"
}).round(2)

print("\nPesticide Usage Group Performance:")
print(risk_analysis)
```

```python
# Weather & Environmental Conditions Analysis

weather_analysis = df[[
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Humidity_pct",
    "Yield_Tonnes_Ha"
]].corr().round(2)

print("Correlation between Weather Factors and Yield:")
print(weather_analysis)

print("\nRainfall-Yield Correlation:",
      weather_analysis.loc["Rainfall_mm", "Yield_Tonnes_Ha"])

print("Temperature-Yield Correlation:",
      weather_analysis.loc["Avg_Temperature_C", "Yield_Tonnes_Ha"])

print("Humidity-Yield Correlation:",
      weather_analysis.loc["Humidity_pct", "Yield_Tonnes_Ha"])
```

```python
# Weather & Environmental Conditions Analysis

weather_analysis = df[[
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Humidity_pct",
    "Yield_Tonnes_Ha"
]].corr().round(2)

print("Correlation between Weather Factors and Yield:")
print(weather_analysis)

print("\nRainfall-Yield Correlation:",
      weather_analysis.loc["Rainfall_mm", "Yield_Tonnes_Ha"])

print("Temperature-Yield Correlation:",
      weather_analysis.loc["Avg_Temperature_C", "Yield_Tonnes_Ha"])

print("Humidity-Yield Correlation:",
      weather_analysis.loc["Humidity_pct", "Yield_Tonnes_Ha"])
```

```python
# Seed Quality Analysis

print("Average Seed Quality Score:",
      round(df["Seed_Quality_Score"].mean(), 2))

print("Average Yield (tonnes/ha):",
      round(df["Yield_Tonnes_Ha"].mean(), 2))

print("Average Profit (INR):",
      round(df["Profit_INR"].mean(), 2))

seed_analysis = df[
    ["Seed_Quality_Score", "Yield_Tonnes_Ha", "Profit_INR"]
].corr().round(2)

print("\nCorrelation between Seed Quality and Performance:")
print(seed_analysis)

print("\nSeed Quality-Yield Correlation:",
      seed_analysis.loc["Seed_Quality_Score", "Yield_Tonnes_Ha"])

print("Seed Quality-Profit Correlation:",
      seed_analysis.loc["Seed_Quality_Score", "Profit_INR"])
```

```python
# Farm Area Analysis

print("Average Farm Area (hectares):",
      round(df["Farm_Area_Hectares"].mean(), 2))

print("Average Production (tonnes):",
      round(df["Production_Tonnes"].mean(), 2))

print("Average Revenue (INR):",
      round(df["Revenue_INR"].mean(), 2))

print("Average Profit (INR):",
      round(df["Profit_INR"].mean(), 2))

farm_area_analysis = df[[
    "Farm_Area_Hectares",
    "Production_Tonnes",
    "Revenue_INR",
    "Profit_INR"
]].corr().round(2)

print("\nCorrelation between Farm Area and Performance:")
print(farm_area_analysis)

print("\nFarm Area-Production Correlation:",
      farm_area_analysis.loc["Farm_Area_Hectares", "Production_Tonnes"])

print("Farm Area-Revenue Correlation:",
      farm_area_analysis.loc["Farm_Area_Hectares", "Revenue_INR"])

print("Farm Area-Profit Correlation:",
      farm_area_analysis.loc["Farm_Area_Hectares", "Profit_INR"])
```

# Findings, Conclusion & Recommendations

## Key Findings

1. Kharif season showed the highest average yield and average profit among the three seasons.
2. Sugarcane showed the highest average yield and average profit among the crops.
3. Punjab showed the highest average profit among the states.
4. Drip irrigation provided the highest average yield.
5. Rainfed irrigation showed the highest water efficiency.
6. Moderate fertilizer usage showed better average yield compared with very high fertilizer usage.
7. Lower pesticide usage groups showed higher disease/pest risk in this dataset.
8. Weather and soil factors showed weak correlation with yield in the dataset.
9. Market price and revenue showed a positive relationship.
10. Revenue and profit showed a strong positive relationship.

## Conclusion

The analysis shows that agricultural performance varies across seasons, crops, states, irrigation methods and resource usage. Kharif season performed better in terms of yield and profit. Crop selection and irrigation method also had an important effect on agricultural performance. The analysis of resources such as fertilizer, pesticide and water helped identify variations in productivity and efficiency.

## Recommendations

- Focus on suitable crops and practices during high-performing seasons.
- Consider efficient irrigation methods such as drip irrigation for better yield.
- Promote water-efficient practices where appropriate.
- Avoid excessive fertilizer usage and follow balanced application.
- Monitor disease and pest risks carefully.
- Use market price and cost information while planning crop production.
- Use the identified seasonal and regional patterns for better agricultural planning.
